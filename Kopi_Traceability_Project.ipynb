```python
import os
import shutil
import random

# Tentukan sumber asli setelah unzip
sumber_asli = '/content/dataset'
if 'dataset_kopi' in os.listdir(sumber_asli):
    sumber_asli = '/content/dataset/dataset_kopi'

# Tentukan folder tujuan yang rapi
folder_final = '/content/dataset_siap_pakai'
if os.path.exists(folder_final):
    shutil.rmtree(folder_final)

kategori_kopi = [d for d in os.listdir(sumber_asli) if os.path.isdir(os.path.join(sumber_asli, d))]

for cat in kategori_kopi:
    nama_bersih = cat.replace('Damange', 'Damage') # Memperbaiki typo
    os.makedirs(os.path.join(folder_final, 'train', nama_bersih), exist_ok=True)
    os.makedirs(os.path.join(folder_final, 'val', nama_bersih), exist_ok=True)

    semua_foto = [f for f in os.listdir(os.path.join(sumber_asli, cat)) if f.lower().endswith(('.png', '.jpg', '.jpeg'))]
    random.shuffle(semua_foto)

    titik_pisah = int(0.8 * len(semua_foto))
    for i, f in enumerate(semua_foto):
        target = 'train' if i < titik_pisah else 'val'
        shutil.copy(os.path.join(sumber_asli, cat, f), os.path.join(folder_final, target, nama_bersih, f))

print(f"✅ Data siap di: {folder_final}")
```
